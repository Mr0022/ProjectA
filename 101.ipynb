```python
import optuna

dst_storage = "sqlite:///all_studies.db"

sources = [
    ("LSTMM1", "sqlite:///LSTM results/LSTMM1/study.db"),
    ("LSTMM5",   "sqlite:///LSTM results/LSTMM5/study.db"),
    ("LSTMM22",  "sqlite:///LSTM results/LSTMM22/study.db"),
    ("EVENTTCN1",  "sqlite:///LSTM results/EVENTTCN1/study.db"),
    ("EVENTTCN5",  "sqlite:///LSTM results/EVENTTCN5/study.db"),
    ("EVENTTCN22",  "sqlite:///LSTM results/EVENTTCN22/study.db"),
    ("ModernTCN1",  "sqlite:///LSTM results/ModernTCN1/study.db"),
    ("ModernTCN5",  "sqlite:///LSTM results/ModernTCN5/study.db"),
    ("ModernTCN22",  "sqlite:///LSTM results/ModernTCN22/study.db"),
]

for study_name, src_storage in sources:
    src = optuna.load_study(
        study_name=study_name,
        storage=src_storage
    )

    dst = optuna.create_study(
        study_name=study_name,
        storage=dst_storage,
        load_if_exists=True,
        direction=src.direction,
    )

    dst.add_trials(src.trials)

    print(
        f"{study_name}: copied {len(src.trials)} trials"
    )
```
